```python
import pandas as pd
import numpy as np
import datetime as dt
import yfinance as yf

def get_stock_data(ticker, start_date, end_date, filename):
    
    # Download stock data
    stock = yf.Ticker(ticker)
    stock_df = stock.history(
        start=start_date,
        end=end_date
    )
    
    # Reset Date index
    stock_df = stock_df.reset_index()
    
    # Remove timezone from Date
    stock_df['Date'] = pd.to_datetime(
        stock_df['Date']
    ).dt.tz_localize(None)
    
    # Reset index
    stock_df.reset_index(drop=True, inplace=True)
    
    # Export to CSV
    stock_df.to_csv(filename, index=False)
    
    return stock_df
```

```python
import pandas as pd
import numpy as np
import datetime as dt
import yfinance as yf
```

```python
startdate = dt.datetime(2021, 1, 1)
enddate = dt.datetime(2026, 9, 21)
```

```python
def get_stock_data(ticker, start_date, end_date, filename):
    
    stock = yf.Ticker(ticker)
    
    stock_df = stock.history(
        start=start_date,
        end=end_date
    )
    
    stock_df = stock_df.reset_index()
    
    stock_df['Date'] = pd.to_datetime(
        stock_df['Date']
    ).dt.tz_localize(None)
    
    stock_df.reset_index(drop=True, inplace=True)
    
    stock_df.to_csv(filename, index=False)
    
    return stock_df
```

```python
stocks = {
    "TATAELXSI": "TATAELXSI.NS",
    "HDFCBANK": "HDFCBANK.NS",
    "SUNPHARMA": "SUNPHARMA.NS",
    "LT": "LT.NS",
    "NTPC": "NTPC.NS",
    "HAL": "HAL.NS",
    "TITAN": "TITAN.NS",
    "BHARTIARTL": "BHARTIARTL.NS",
    "MARUTI": "MARUTI.NS",
    "ITC": "ITC.NS"
}
```

```python
stock_data = {}

for name, ticker in stocks.items():
    
    filename = f"{name}_stock_data.csv"
    
    stock_data[name] = get_stock_data(
        ticker,
        startdate,
        enddate,
        filename
    )
    
    print(f"{name} downloaded successfully")
```
